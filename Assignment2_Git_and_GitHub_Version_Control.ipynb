## 4. Notebook Creation

**4.1** Create `assignment2.ipynb` in the cloned repository.

**4.2** Include the required Markdown cell.

**4.3** Add the multiplication function.

```python
import nbformat

nb = nbformat.v4.new_notebook()
nb.cells = [
    nbformat.v4.new_markdown_cell(
        "# Git and GitHub Assignment 2\n"
        "This notebook demonstrates basic Git operations."
    ),
    nbformat.v4.new_code_cell(
        "def multiply_numbers(a, b):\n"
        "    return a * b\n\n"
        "print(multiply_numbers(3, 4))"
    )
]
nbformat.write(nb, "assignment2.ipynb")
print("Notebook created:", "assignment2.ipynb")
!ls -l assignment2.ipynb
```

```python
def multiply_numbers(a, b):
    return a * b

print("Multiplication result:", multiply_numbers(3, 4))
```

```python
from google.colab import userdata
import os
import base64

token = userdata.get("assignment2-rkv1")
assert token, "Add your PAT in Colab Secrets under assignment2-rkv1."
auth = base64.b64encode(f"x-access-token:{token}".encode()).decode()
os.environ["GIT_AUTH_HEADER"] = f"AUTHORIZATION: basic {auth}"
del token, auth

!git -c http.extraheader="$GIT_AUTH_HEADER" push origin main
```

```python
import nbformat

nb = nbformat.read("assignment2.ipynb", as_version=4)
division_code = (
    "def divide_numbers(a, b):\n"
    "    if b != 0:\n"
    "        return a / b\n"
    "    else:\n"
    "        return 'Cannot divide by zero'\n\n"
    "print(divide_numbers(10, 2))"
)
nb.cells.append(nbformat.v4.new_code_cell(division_code))
nbformat.write(nb, "assignment2.ipynb")
print("Division function added to assignment2.ipynb")
```

```python
def divide_numbers(a, b):
    if b != 0:
        return a / b
    else:
        return "Cannot divide by zero"

print("Division result:", divide_numbers(10, 2))
print("Zero-division test:", divide_numbers(10, 0))
```

**Expected outputs:** `5.0` and `Cannot divide by zero`.
