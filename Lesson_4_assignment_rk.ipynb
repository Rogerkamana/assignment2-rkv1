```python
# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Your code here:
try:
    df1 = pd.DataFrame({
        "ID": [1, 2],
        "Name": ["Alice", "Bob"]
    })

    df2 = pd.DataFrame({
        "ID": [3, 4],
        "Name": ["Charlie", "David"]
    })

    df3 = pd.DataFrame({
        "ID": [1, 2, 3, 4],
        "Department": ["HR", "IT", "Finance", "Marketing"]
    })

    # Vertical Concatenation
    combined = pd.concat([df1, df2], ignore_index=True)

    print("Concatenated DataFrame:")
    print(combined)

    # Merge horizontally using ID
    merged = pd.merge(combined, df3, on="ID")

    print("\nMerged DataFrame:")
    print(merged)

    print("\nShape:", merged.shape)

except Exception as e:
    print("Error:", e)

```

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
try:
    class EvenNumbers:
        def __init__(self):
            self.num = 0
            self.count = 0

        def __iter__(self):
            return self

        def __next__(self):
            if self.count >= 5:
                raise StopIteration

            value = self.num
            self.num += 2
            self.count += 1
            return value

    print("Iterator Output:")
    even_iter = EvenNumbers()

    for num in even_iter:
        print(num)

    # Generator
    def even_generator():
        for i in range(0, 10, 2):
            yield i

    print("\nGenerator Output:")
    for num in even_generator():
        print(num)

except Exception as e:
    print("Error:", e)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
from functools import reduce

try:
    nums = [1, 2, 3, 4, 5]

    squares = list(map(lambda x: x**2, nums))

    product = reduce(lambda x, y: x * y, nums)

    even_numbers = list(filter(lambda x: x % 2 == 0, nums))

    print("Original List:", nums)
    print("Squared Numbers:", squares)
    print("Product:", product)
    print("Even Numbers:", even_numbers)

except Exception as e:
    print("Error:", e)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
try:
    class Rectangle:
        def __init__(self, length, width):
            self.length = length
            self.width = width

        def area(self):
            return self.length * self.width

        def perimeter(self):
            return 2 * (self.length + self.width)

    rect1 = Rectangle(10, 5)
    rect2 = Rectangle(8, 4)

    print("Rectangle 1")
    print("Area:", rect1.area())
    print("Perimeter:", rect1.perimeter())

    print("\nRectangle 2")
    print("Area:", rect2.area())
    print("Perimeter:", rect2.perimeter())

except Exception as e:
    print("Error:", e)
```

```python
# Task 12: Pandas Data Analysis
# Using the following DataFrame, perform these tasks:
# 1. Find the average salary by department
# 2. Get the names of employees with salary > 60000
# 3. Add a new column 'Bonus' which is 10% of salary

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# Your code here:
try:
    employees = {
        "Name": ["John", "Emma", "Michael", "Sophia", "James"],
        "Department": ["HR", "IT", "IT", "Finance", "HR"],
        "Salary": [55000, 70000, 65000, 80000, 60000]
    }

    emp_df = pd.DataFrame(employees)

    print("Employee Data:")
    print(emp_df)

    # Average salary by department
    avg_salary = emp_df.groupby("Department")["Salary"].mean()

    print("\nAverage Salary by Department:")
    print(avg_salary)

    # Salary > 60000
    high_salary = emp_df[emp_df["Salary"] > 60000]["Name"]

    print("\nEmployees with Salary > 60000:")
    print(high_salary.tolist())

    # Bonus column
    emp_df["Bonus"] = emp_df["Salary"] * 0.10

    print("\nDataFrame with Bonus Column:")
    print(emp_df)

except Exception as e:
    print("Error:", e)
```
